```python
import json, pandas as pd
frames = pd.DataFrame(json.load(open("frame.json")))
frames.head(5)
```

```python
frames = frames.drop(columns=["product_code",  "compare_at", "site_category", "shipping_grams", "spec_notes", "variant_id", "options"])
frames = frames.rename(columns={"variant_title": "size (cm)"})

frames.head()
```

```python
frames.head(2)
```

```python
frames["name"] = frames["name"].str.extract(r"^(.*) - .+$")
frames.head()
```

```python
frames = (
    frames
    .groupby(["name", "image_url", "source_url"], as_index=False)
    .agg({
        "brand": "first",
        "component_type": "first",
        "price": "first",
        "currency": "first",
        "weight_grams": "first",
        "description": "first",
        "frame_brake_drilled": "first",
        "fork_brake_drilled": "first",
        "specs": "first",
        "size (cm)": list,
    })
    .rename(columns={"size (cm)": "sizes"})
)
frames.head()
```

```python
import re

def tire_clearance(specs):
    text = " ".join(v for k, v in specs.items() if "tire" in k.lower() or "clear" in k.lower())
    nums = [int(n) for n in re.findall(r"(\d{2,3})\s*mm", text)]
    return min(nums) if nums else None

frames["max_tire_clearance_mm"] = frames["specs"].apply(tire_clearance)
frames[["name", "max_tire_clearance_mm"]].head(10)
```
